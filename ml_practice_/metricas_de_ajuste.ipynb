```python
import pandas as pd

url = "/content/Dados Comunidade (respostas) - dados (2).csv"

df = pd.read_csv(url)
df.head()
```

# **Prever se uma pessoa é feliz ou não**

```python
df = df.replace({'Sim':1,'Não':0})
```

```python
df.head()
```

```python
# Tratar variáveis para que sejam numéricas

pd.get_dummies(df[['Como conheceu o Téo Me Why?']])
```

# **Transformar em Dummies**

```python
dummy_vars = ['Como conheceu o Téo Me Why?',
'Quantos cursos acompanhou do Téo Me Why?',
'Estado que mora atualmente',
'Área de Formação',
'Tempo que atua na área de dados',
'Posição da cadeira (senioridade)']

df_dummy = pd.get_dummies(df[dummy_vars]).astype(int)
```

```python
df_dummy.head()
```

# **Adicionar as que já eram numéricas**

```python
num_vars = [
    'Curte games?',
    'Curte futebol?',
    'Curte jogos de tabuleiro?',
    'Curte jogos de fórmula 1?',
    'Curte jogos de MMA?',
    'Idade'
]

df_analise = pd.get_dummies(df[dummy_vars]).astype(int)
df_analise[num_vars] = df[num_vars].copy()
```

```python
df_analise.head()
```

```python
df_analise['Pessoa Feliz'] = df['Você se considera uma pessoa feliz?'].copy()
```

```python
df_analise.head()
```

```python
from sklearn import tree
arvore = tree.DecisionTreeClassifier(random_state=42,
                                    min_samples_leaf=5) # pelo menos 5 folhas
```

```python
features = df_analise.columns[:-1].tolist()
features
```

```python
X = df_analise[features]
y = df_analise['Pessoa Feliz']
```

```python
arvore.fit(X, y)
```

```python
# Predição

arvore_predict = arvore.predict(X)
arvore_predict
```

```python
# df original

df_predict = df_analise[['Pessoa Feliz']]
df_predict['predict_arvore'] = arvore_predict
df_predict
```

```python
(df_predict['Pessoa Feliz'] == df_predict['predict_arvore']).mean()

```

```python
# Acertando 86% - Acurácia
# Matriz de Confusão: Como descobrir onde estão os erros?

pd.crosstab(df_predict['Pessoa Feliz'], df_predict['predict_arvore'])
```

# **Como plotar a curva ROC?**

```python
from sklearn import metrics
```

```python
# Acurácia
acc_arvore = metrics.accuracy_score(df_predict['Pessoa Feliz'], df_predict['predict_arvore'])
acc_arvore
```

```python
# Precisão
precisao_arvore = metrics.precision_score(df_predict['Pessoa Feliz'], df_predict['predict_arvore'])
precisao_arvore
```

```python
# Recall
recall_arvore = metrics.recall_score(df_predict['Pessoa Feliz'], df_predict['predict_arvore'])
recall_arvore
```

```python
# Especificidade
especificidade_arvore = metrics.recall_score(df_predict['Pessoa Feliz'], df_predict['predict_arvore'], pos_label=0)
especificidade_arvore
```

```python
df_predict['proba_arvore'] = arvore.predict_proba(X)[:, 1]
df_predict.head()
```

```python
# Curva ROC
roc = metrics.roc_curve(df_predict['Pessoa Feliz'],df_predict['proba_arvore'])
```

```python
import matplotlib.pyplot as plt

plt.plot(roc[0], roc[1])
plt.grid(True)
plt.title('ROC Curve')
plt.xlabel('1 - Especificidade')
plt.ylabel('Recall')
plt.figure(dpi=400)
```

```python
plt.hist(df_predict['proba_arvore'])
```

# **Calcular a área da curva ROC**

```python
auc_arvore = metrics.roc_auc_score(df_predict['Pessoa Feliz'],df_predict['proba_arvore'])
auc_arvore
```

```python
plt.plot(roc[0], roc[1], label=f"Árvore: {auc_arvore:.2f}")
plt.grid(True)
plt.legend()
plt.title('ROC Curve')
plt.xlabel('1 - Especificidade')
plt.ylabel('Recall')
plt.figure(dpi=400)

# Quanto você consegue crescer o recall sem penalizar a especificidade
```

# **Naive Bayes**

```python
from sklearn import naive_bayes

# Outro Modelo:

naive = naive_bayes.GaussianNB()
naive.fit(X, y)
```

```python
# Predição:
naive_predict = naive.predict(X)

# Coluna com a Predição:
df_predict_naive = df_analise[['Pessoa Feliz']]
df_predict_naive['predict_naive'] = naive_predict
```

```python
# Matriz de Confusão: Como descobrir onde estão os erros?

pd.crosstab(df_predict_naive['Pessoa Feliz'], df_predict_naive['predict_naive'])
```

```python
# Métricas para a Árvore de Decisão
acc_arvore = metrics.accuracy_score(df_predict['Pessoa Feliz'], df_predict['predict_arvore'])
precisao_arvore = metrics.precision_score(df_predict['Pessoa Feliz'], df_predict['predict_arvore'])
recall_arvore = metrics.recall_score(df_predict['Pessoa Feliz'], df_predict['predict_arvore'])
especificidade_arvore = metrics.recall_score(df_predict['Pessoa Feliz'], df_predict['predict_arvore'], pos_label=0)
df_predict['proba_arvore'] = arvore.predict_proba(X)[:, 1]
roc_arvore = metrics.roc_curve(df_predict['Pessoa Feliz'], df_predict['proba_arvore'])
auc_arvore = metrics.roc_auc_score(df_predict['Pessoa Feliz'], df_predict['proba_arvore'])

print(f"--- Métricas da Árvore de Decisão ---")
print(f"Acurácia: {acc_arvore:.2f}")
print(f"Precisão: {precisao_arvore:.2f}")
print(f"Recall: {recall_arvore:.2f}")
print(f"Especificidade: {especificidade_arvore:.2f}")
print(f"AUC: {auc_arvore:.2f}")
print("\n")

# Métricas para o Naive Bayes
acc_naive = metrics.accuracy_score(df_predict_naive['Pessoa Feliz'], df_predict_naive['predict_naive'])
precisao_naive = metrics.precision_score(df_predict_naive['Pessoa Feliz'], df_predict_naive['predict_naive'])
recall_naive = metrics.recall_score(df_predict_naive['Pessoa Feliz'], df_predict_naive['predict_naive'])
especificidade_naive = metrics.recall_score(df_predict_naive['Pessoa Feliz'], df_predict_naive['predict_naive'], pos_label=0)
df_predict_naive['proba_naive'] = naive.predict_proba(X)[:, 1]
roc_naive = metrics.roc_curve(df_predict_naive['Pessoa Feliz'], df_predict_naive['proba_naive'])
auc_naive = metrics.roc_auc_score(df_predict_naive['Pessoa Feliz'], df_predict_naive['proba_naive'])

print(f"--- Métricas do Naive Bayes ---")
print(f"Acurácia: {acc_naive:.2f}")
print(f"Precisão: {precisao_naive:.2f}")
print(f"Recall: {recall_naive:.2f}")
print(f"Especificidade: {especificidade_naive:.2f}")
print(f"AUC: {auc_naive:.2f}")
print("\n")

# Curva ROC para ambos os modelos
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6), dpi=400)
plt.plot(roc_arvore[0], roc_arvore[1], label=f"Árvore de Decisão: {auc_arvore:.2f}")
plt.plot(roc_naive[0], roc_naive[1], label=f"Naive Bayes: {auc_naive:.2f}", linestyle='--')
plt.grid(True)
plt.legend()
plt.title('Curva ROC Comparativa')
plt.xlabel('1 - Especificidade (Taxa de Falsos Positivos)')
plt.ylabel('Recall (Taxa de Verdadeiros Positivos)')
plt.show()
```
